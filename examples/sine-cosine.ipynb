```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 500)
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(x, np.sin(x), label="sin(x)", color="#2d698f", linewidth=2)
ax.plot(x, np.cos(x), label="cos(x)", color="#d66b2c", linewidth=2)
ax.set(title="Sine and cosine", xlabel="x (radians)", ylabel="Amplitude", xlim=(0, 2 * np.pi))
ax.axhline(0, color="gray", linewidth=0.7)
ax.grid(True, alpha=0.25)
ax.legend()
fig.tight_layout()
plt.show()
```
