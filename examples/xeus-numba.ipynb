# Numba in the browser
Run in the xeus-python environment. This checks real nopython compilation, not an interpreted fallback. The first call includes compilation time.

Qrisp is not included: its full dependency graph is not currently available in the selected WASM channels.

```python
import numpy as np
import numba, llvmlite
from numba import njit

@njit
def sum_squares(values):
    result = 0.0
    for value in values:
        result += value * value
    return result

values = np.arange(10, dtype=np.float64)
assert sum_squares(values) == 285.0
assert len(sum_squares.nopython_signatures) == 1
print('Numba:', numba.__version__, 'llvmlite:', llvmlite.__version__)
print('Compiled result:', sum_squares(values))
print('Nopython signatures:', len(sum_squares.nopython_signatures))
```

```python
import matplotlib.pyplot as plt
sizes = np.arange(1, 11)
results = [sum_squares(np.arange(n, dtype=np.float64)) for n in sizes]
plt.bar(sizes, results)
plt.xlabel('Number of values')
plt.ylabel('Sum of squares (compiled with Numba)')
plt.title('Real WASM compilation in xeus-python')
plt.show()
```
